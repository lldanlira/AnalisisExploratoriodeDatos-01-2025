## 5. Rendimientos Acumulados

Los rendimientos simples acumulados hasta $T$ se obtienen por composición geométrica:

$$
R_T^{(s)}=\prod_{t=1}^{T}(1+r_t)-1
=\frac{P_T}{P_0}-1.
$$

Los log-rendimientos son aditivos, por lo que el log-rendimiento acumulado es:

$$
G_T=\sum_{t=1}^{T}g_t
=\ln\left(\frac{P_T}{P_0}\right).
$$

$G_T$ está expresado en unidades logarítmicas y **no debe compararse directamente como porcentaje** con $R_T^{(s)}$. Para convertirlo al rendimiento simple equivalente:

$$
R_T^{(\log\to s)}=e^{G_T}-1.
$$

Por identidad matemática, $R_T^{(\log\to s)}=R_T^{(s)}$, salvo pequeñas diferencias numéricas de redondeo.


````python
# Rendimiento Acumulado Simple

import matplotlib.pyplot as plt
cum_returns = (1 + returns_simple).cumprod()-1

plt.figure(figsize=(12,8))
cum_returns.plot()
plt.title("Rendimiento Acumulado")
plt.show()

printmd("Rendimiento Acumulado","E2")
display(cum_returns.tail(1))
# En porcentaje
printmd("Rendimiento Acumulado Porcentaje %","E2")
display(cum_returns.tail(1)*100)
````

````python
import plotly.express as px

cum_log_returns = returns_log.cumsum()

fig = px.line(cum_log_returns, x=cum_log_returns.index, y=cum_log_returns.columns)
fig.update_layout(title='Rendimiento Logarítmico Acumulado', xaxis_title='Fecha', yaxis_title='Rendimiento Logarítmico Acumulado')
fig.show()

printmd("Rendimiento Logarítmico Acumulado","E2")

display(cum_log_returns.tail(1))
````

````python
# Conversión a rendimientos simples
rendimiento_simple_acumulado_logaritmico = np.exp(cum_log_returns) - 1
# Resumen
printmd('Rendimiento simple acumulado',"E2")
display(cum_returns.tail(1))
printmd('Rendimiento logarítmico acumulado',"E2")
display(cum_log_returns.tail(1))
printmd('Rendimiento simple acumulado a partir de log',"E2")
display(rendimiento_simple_acumulado_logaritmico.tail(1))
````

### Comprobación de equivalencia

La siguiente prueba debe producir un error cercano a cero:

$$
\max\left|R_T^{(s)}-\left(e^{G_T}-1\right)\right|\approx 0.
$$


````python
error_maximo = (returns_log - returns_simple).abs().to_numpy().max()
printmd(f'Error máximo entre ambos métodos: {error_maximo:.3e}',"E2")
assert np.allclose(cum_returns, rendimiento_simple_acumulado_logaritmico, rtol=1e-10, atol=1e-12)
````

## Paso 8 — Covarianza


La **covarianza** mide cómo dos variables (activos financieros) se mueven **conjuntamente** respecto a sus medias.

* Si es **positiva** → se mueven en la misma dirección
* Si es **negativa** → se mueven en direcciones opuestas
* Si es **cercana a 0** → no hay relación lineal clara


### 📐 Fórmula de la Covarianza

$$
\text{Cov}(X,Y) =
\frac{1}{T-1}
\sum_{t=1}^{T}
(X_t - \mu_X)(Y_t - \mu_Y)
$$

Donde:

* $\text{Cov}(X,Y)$ = Covarianza entre activos
* $\mu_X, \mu_Y$ = Medias
* $T$ = Número de observaciones


---

### 🎯 Interpretación financiera

* Covarianza **positiva alta** → activos suben/bajan juntos
* Covarianza **negativa** → uno sube mientras el otro baja → **diversificación**
* Covarianza **cerca de 0** → movimientos independientes

⚠️ La covarianza **depende de la escala**, por eso en finanzas se usa más la **correlación**:



````python
cov_matrix = returns_log.cov()
printmd("Matriz de Covarianza:","E2")
display(cov_matrix)
````


## Paso 9 — Correlación



La **correlación** mide el grado de relación lineal entre dos activos financieros.
Indica si los rendimientos se mueven **en la misma dirección**, **en dirección opuesta** o **sin relación**.



### 📐 Fórmula de correlación de Pearson


$$
\rho_{X,Y} =
\frac{\text{Cov}(X,Y)}{\sigma_X \sigma_Y}
$$


Donde:

* $\rho_{X,Y}$ = Correlación entre activos
* $\text{Cov}(X,Y)$ = Covarianza
* $\sigma_X, \sigma_Y$ = Desviaciones estándar

---

### 📊 Interpretación

| Correlación   | Significado                |
| ------------- | -------------------------- |
| **+1**        | Relación perfecta positiva |
| **0**         | Sin relación               |
| **−1**        | Relación perfecta negativa |
| ---- | ---- |
| **0.7 a 1**   | Alta correlación           |
| **0.3 a 0.7** | Correlación media          |
| **0 a 0.3**   | Baja correlación           |


---

### 🎯 Interpretación financiera

* Correlación **alta positiva** → activos se mueven juntos → **menos diversificación**
* Correlación **cercana a 0** → activos independientes → **mejor diversificación**
* Correlación **negativa** → activos se compensan → **reducción de riesgo**

---

### 📌 Uso en portafolios

La matriz de correlación se usa para:

* Construir portafolios eficientes (Markowitz)
* Reducir riesgo total
* Identificar activos redundantes
* Optimizar diversificación




````python

corr = returns_log.corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Matriz de Correlación")
plt.show()
````
